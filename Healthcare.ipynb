```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#Global plotting style
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"]=(8,5)

#Display options
pd.set_option("display.max_columns",None)
pd.set_option("display.width",140)

print("✅ Libraries imported successfully")
```

```python
#Load the dataset
df=pd.read_csv("healthcare_dataset.csv")

print("✅dataset loaded successfully")
print(f"Shape:{df.shape[0]} rows X {df.shape[1]} columns")
df.head()
```

```python
# ---- First 5 rows ----
print("=== HEAD ===")
display(df.head())

# ---- Last 5 rows ----
print("\n=== TAIL ===")
display(df.tail())
```

```python
# ---- shape ----
print("\n === shape ===")
display(f" Rows :df.{df.shape[0]}, Columns :{df.shape[1]}")

# ---- columns ----
print("\n=== columns ===")
display(df.columns.tolist())
```

```python
# ---- info ----
print("\n=== INFO ===")
df.info()
# ---- Describe (only numeric works by default) ----
print("\n=== DESCRIBE (numerical) ===")
display(df.describe())

# ---- Describe ALL (categorical too) ----
print("\n=== DESCRIBE (all) ===")
display(df.describe(include="all"))

# head() / tail() → peek first/last rows
# shape → dimensions
# info() → dtypes + non-null counts
# describe() → stats (numeric by default)
# describe(include="all") → includes categoricals
```

```python
# ---- Missing values ----
print("=== Missing values per column ===")
print(df.isnull().sum())

total_missing = df.isnull().sum().sum()
if total_missing == 0:
    print("\n✅ No missing values — no imputation required.")
else:
    print(f"\n⚠️ Total missing: {total_missing} — applying imputation")
    for col in df.select_dtypes(include=np.number).columns:
        df[col] = df[col].fillna(df[col].median())
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].fillna(df[col].mode()[0])
    print("✅ Missing values imputed")    
```

```python
# ---- Duplicate rows ----
dup_count = df.duplicated().sum()
print(f"Duplicate rows: {dup_count}")

if dup_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print(f"✅ {dup_count} duplicates removed")
    print(f"New shape: {df.shape}")
else:
    print("✅ No duplicates found")
```

```python
# ---- Convert date columns to datetime ----
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
df["Discharge Date"]    = pd.to_datetime(df["Discharge Date"])

print("✅ Date columns converted:")
print(df[["Date of Admission", "Discharge Date"]].dtypes)
```

```python
# ---- Clean the Name column ----
# Original names look like: "Bobby JacksOn", "MS. nAtalIE gAMble", "MR. TYler TAYLOR Phd"
# Steps: remove titles/prefixes, then title-case the rest.

titles = r"^(mr\.|mrs\.|ms\.|dr\.)\s+"
suffixes = r"\s+(md|phd|dds|dvm|jr\.|sr\.|ii|iii|iv|v)$"

df["Name"] = (
    df["Name"]
    .str.replace(titles, "", regex=True, case=False)   # remove prefix
    .str.replace(suffixes, "", regex=True, case=False)  # remove suffix
    .str.strip()
    .str.title()                                        # Title Case
)

print("✅ Names cleaned. Sample:")
display(df["Name"].head(10))
```

```python
# ---- Check for negative billing amounts (data errors) ----
neg_billing = df[df["Billing Amount"]<0]

print(f"Negative billing rows found: {len(neg_billing)}")
print("\nExamples:")
display(neg_billing[["Name", "Medical Condition", "Billing Amount"]].head())
```

```python
# ---- Drop negative billing rows ----
before = len(df)
df = df[df["Billing Amount"] >= 0].reset_index(drop=True)
after = len(df)

print(f"✅ Dropped {before - after} rows with negative billing")
print(f"New shape: {df.shape}")
```

```python
df["Hospital"]=df["Hospital"].str.replace(",","",regex=False).str.strip()

print("✅ Hospital names cleaned. Sample:")
display(df["Hospital"].head(5))
```

```python
# ---- Outlier detection using IQR (Billing Amount) ----
col = "Billing Amount"
q1 = df[col].quantile(0.25)
q3 = df[col].quantile(0.75)

iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper= q3 + 1.5 * iqr

outliers=df[(df[col]<lower) | (df[col] > upper)]

print("=== Billing Amount IQR ===")
print(f"Q1: {q1:.2f}")
print(f"Q3: {q3:.2f}")
print(f"IQR: {iqr:.2f}")
print(f"Lower limit: {lower:.2f}")
print(f"Upper limit: {upper:.2f}")
print(f"Number of outliers: {len(outliers)}")
```

```python
# Feature 1: Length of stay (days)
df["length_of_stay"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days

print("✅ length_of_stay created")
print(df["length_of_stay"].describe().round(2))
```

```python
# Feature 2: Admission year and month (from date)
df["admission_year"]  = df["Date of Admission"].dt.year
df["admission_month"] = df["Date of Admission"].dt.month

print("✅ admission_year and admission_month created")
display(df[["Date of Admission", "admission_year", "admission_month"]].head())
```

```python
# Feature 3: Age groups
df["age_group"] = pd.cut(
    df["Age"],
    bins=[-1, 18, 35, 50, 65, 120],
    labels=["0-18", "19-35", "36-50", "51-65", "66+"]
)

print("✅ age_group created")
print(df["age_group"].value_counts())
```

```python
# Feature 4: Billing categories
df["billing_category"] = pd.cut(
    df["Billing Amount"],
    bins=[-1, 10000, 25000, 40000, float("inf")],
    labels=["Low", "Medium", "High", "Very High"]
)

print("✅ billing_category created")
print(df["billing_category"].value_counts())
```

```python
# Feature 5: Binary emergency flag using map()
df["is_emergency"] = df["Admission Type"].map({"Emergency": 1, "Urgent": 1, "Elective": 0})

print("✅ is_emergency created via map()")
print(df["is_emergency"].value_counts())
```

```python
# ---- Descriptive statistics ----
print("=== Numeric Summary ===")
display(df[["Age", "Billing Amount", "length_of_stay"]].describe().round(2))

# ---- Frequencies ----
print("\n=== Gender ===")
print(df["Gender"].value_counts())

print("\n=== Admission Type ===")
print(df["Admission Type"].value_counts())

print("\n=== Medical Condition ===")
print(df["Medical Condition"].value_counts())

print("\n=== Test Results ===")
print(df["Test Results"].value_counts())

print("\n=== Insurance Provider ===")
print(df["Insurance Provider"].value_counts())
```

```python
# ---- Group-wise analysis: Avg billing by condition ----
print("=== Average Billing by Medical Condition ===")
print(df.groupby("Medical Condition", observed=True)["Billing Amount"].mean().round(2).sort_values(ascending=False))

print("\n=== Average Billing by Admission Type ===")
print(df.groupby("Admission Type", observed=True)["Billing Amount"].mean().round(2))

print("\n=== Average Length of Stay by Admission Type ===")
print(df.groupby("Admission Type", observed=True)["length_of_stay"].mean().round(2))

print("\n=== Average Billing by Insurance Provider ===")
print(df.groupby("Insurance Provider", observed=True)["Billing Amount"].mean().round(2).sort_values(ascending=False))

print("\n=== Average Billing by Age Group ===")
print(df.groupby("age_group", observed=True)["Billing Amount"].mean().round(2))
```

```python
# ---- Aggregation with multiple stats ----
agg_stats = df.groupby("Admission Type", observed=True)["Billing Amount"].agg(
    ["count", "mean", "median", "min", "max", "std"]
).round(2)

print("=== Billing Statistics by Admission Type ===")
display(agg_stats)
```

```python
# ---- Pivot Table 1: Average billing by Medical Condition × Admission Type ----
pivot1 = pd.pivot_table(
    df,
    index="Medical Condition",
    columns="Admission Type",
    values="Billing Amount",
    aggfunc="mean",
    observed=True
).round(2)

print("=== Pivot 1: Avg Billing by Condition × Admission Type ===")
display(pivot1)
```

```python
# ---- Pivot Table 2: Average length of stay by Age Group × Gender ----
pivot2 = pd.pivot_table(
    df,
    index="age_group",
    columns="Gender",
    values="length_of_stay",
    aggfunc="mean",
    observed=True
).round(2)

print("=== Pivot 2: Avg Length of Stay by Age Group × Gender ===")
display(pivot2)
```

```python
# ---- Correlation ----
numeric_cols = ["Age", "Billing Amount", "length_of_stay", "Room Number", "is_emergency"]
corr_matrix = df[numeric_cols].corr().round(3)

print("=== Correlation Matrix ===")
display(corr_matrix)
```

```python
# Data Visualization
# Matplotlib #1 — Gender distribution bar chart
gender_counts = df["Gender"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(gender_counts.index.astype(str), gender_counts.values, color=["#3498db", "#e91e63"])
plt.title("Gender Distribution of Patients", fontsize=13)
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
for i, v in enumerate(gender_counts.values):
    plt.text(i, v + 20, str(v), ha="center", fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
# Matplotlib #3 — Scatter: Length of Stay vs Billing Amount
plt.figure(figsize=(8, 5))
plt.scatter(df["length_of_stay"], df["Billing Amount"], alpha=0.35, s=10, color="#8e44ad")
plt.title("Length of Stay vs Billing Amount", fontsize=13)
plt.xlabel("Length of Stay (days)")
plt.ylabel("Billing Amount")
plt.tight_layout()
plt.show()
```

```python
# Seaborn #1 — Countplot: Admission Type
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Admission Type", palette="Set2")
plt.title("Admission Type Distribution", fontsize=13)
plt.xlabel("Admission Type")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

```python
# Seaborn #2 — Barplot: Avg billing by Medical Condition
plt.figure(figsize=(9, 5))
sns.barplot(data=df, x="Medical Condition", y="Billing Amount",
            estimator="mean", errorbar=None, palette="Set1")
plt.title("Average Billing by Medical Condition", fontsize=13)
plt.xlabel("Medical Condition")
plt.ylabel("Average Billing Amount")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# Seaborn #3 — Boxplot: Billing by Admission Type
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x="Admission Type", y="Billing Amount", palette="Set3")
plt.title("Billing Amount by Admission Type", fontsize=13)
plt.xlabel("Admission Type")
plt.ylabel("Billing Amount")
plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(7, 5))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1, vmax=1,
    square=True,
    linewidths=0.5
)
plt.title("Correlation Heatmap — Numerical Features", fontsize=13)
plt.tight_layout()
plt.show()
```
